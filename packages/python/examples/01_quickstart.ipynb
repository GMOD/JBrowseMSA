# Quickstart

`MSAView` draws a multiple sequence alignment beside its phylogenetic tree. `msa=` and `tree=` take the document text, a file path, a dict of name to aligned sequence, or a Biopython alignment or `Bio.Phylo` tree. `msa_url=` and `tree_url=` take a URL the viewer fetches.

```python
from msaview import MSAView

DATA = "../../examples/data"

view = MSAView(
    msa=f"{DATA}/globin.aln",
    tree=f"{DATA}/globin.nh",
    hide_header=True,
    height=320,
)
view
```

Setting a trait updates the widget above and keeps its scroll and zoom.

```python
view.color_scheme = "clustal"
```
